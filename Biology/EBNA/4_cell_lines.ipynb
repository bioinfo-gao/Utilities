4 个细胞系在EB 病毒（EBV）致癌机制、EBNA1 驱动的染色体不稳定性（如 11q23 脆性断裂）、三维基因组学（3D Genome）以及表观遗传学研究中，是一组非常经典的实验与对照组合：

一、 4 个细胞系核心特性对比

| 细胞系名称 | 组织/疾病来源 | EBV 感染状态 | 潜伏期/分子表型特征 | 科研与生信分析核心角色 |
| :--- | :--- | :--- | :--- | :--- |
| **DLD1-136** | 人结直肠腺癌（Colorectal Adenocarcinoma） | **EBV 阴性**（外源工程构建） | MSI-H 上皮细胞；通常为 AID 条件性敲除或 Tet-On 诱导表达系 | 非 B 细胞（上皮肿瘤）背景下验证 EBNA1 诱导 DNA 断裂与基因组不稳定性 |
| **GM12878** | 人外周血 B 淋巴细胞（LCL 永生化系） | **EBV 阳性** | **潜伏 III 型（Latency III）**：表达全部 EBNA（1,2,3A-C）与 LMP 蛋白 | **ENCODE Tier 1 金标准细胞系**；三维基因组（Hi-C）、ChIP-seq 和表观组学基准 |
| **Mutu I** | 伯基特淋巴瘤（Burkitt Lymphoma, BL） | **EBV 阳性** | **潜伏 I 型（Latency I）**：仅表达 **EBNA1**、EBERs 及某些 miRNAs | 研究 EBNA1 蛋白单独介导宿主染色体切割（如 11q23）、*MYC* 易位及免疫逃逸的核心模型 |
| **DG75** | 伯基特淋巴瘤（Burkitt Lymphoma, BL） | **EBV 阴性** | 无 EBV 基因组及任何病毒蛋白表达 | **经典的 EBV 阴性 Burkitt 淋巴瘤对照系**；用以排除 B 细胞肿瘤本身背景的干扰 |

二、 联合使用的实验设计逻辑（为什么通常将它们放在一起研究）
在探讨 EBNA1 蛋白对宿主基因组的结合能力、染色体断裂机制以及表观遗传调控 时，这 4 个细胞系构成了一个严密的控制体系：

梯度控制 EBV 病毒蛋白表达（GM12878 vs Mutu I vs DG75）：

DG75（EBV-）：背景对照，证明某种染色体断裂或结合 Peak 不是所有 B 淋巴瘤共有的特征。

Mutu I（EBV+ 潜伏 I 型）：内源性仅高表达 EBNA1，用于观察自然生理状态下 EBNA1 结合宿主 11q23 位点并引发断裂的作用。

GM12878（EBV+ 潜伏 III 型）：同时表达 EBNA2/LMP1 等多种病毒蛋白，可用于对比其他病毒因子是否会增强或拮抗 EBNA1 的作用，并且该细胞系拥有全球最完善的 Hi-C 与 ChIP-seq 生信数据集。

跨组织类型验证（B 细胞 vs 上皮细胞 DLD1-136）：

EBV 不仅引发 B 细胞淋巴瘤，还引发鼻咽癌（NPC）和胃癌等上皮性恶性肿瘤。

使用 DLD1-136（通过工程手段实现 EBNA1 的急控诱导/表达）可以证明：EBNA1 诱导 11q23 等染色体位点断裂是 EBNA1 蛋白本身的普适性生物学效应，而不依赖于 B 淋巴细胞特有成分。
