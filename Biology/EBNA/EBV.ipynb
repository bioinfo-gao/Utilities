EBV（Epstein-Barr 病毒，人类疱疹病毒 4 型，HHV-4）是人类历史上发现的第一种直接致癌病毒（Human Tumor Virus）。
全球超过 90% 的成年人血清呈 EBV 阳性。


![图片描述](./EBV.jpg)

在分子生物学、基因工程与人类基因组学中，EBV 不仅是研究肿瘤发生与病毒潜伏感染的经典模式生物，更是人类建立永生化细胞系与高效染色体外表达载体（Episomal Vectors）的核心工具。

一、 EBV 病毒简介1. 结构与基因组特征病毒分类：属于疱疹病毒科（Herpesviridae）、γ-疱疹病毒亚科（Gammaherpesvirinae）。基因组：双链线状 DNA（dsDNA），基因组全长约 172 kb，编码超过 80 种蛋白质以及丰富的非编码 RNA（如 EBER1/2 和 BART miRNAs）。病毒结构：由内向外依次为 DNA 核心、二十面体核衣壳（Nucleocapsid）、蛋白被膜（Tegument）以及富含糖蛋白刺突（如 gp350、gH/gL）的脂质包膜（Envelope）。2. 细胞嗜性与潜伏感染靶细胞：主要感染人类的 B 淋巴细胞（通过 gp350 结合 CD21 受体）和 上皮细胞。潜伏/裂解双相周期：EBV 感染后极易建立长期潜伏感染（Latency）。在潜伏期，病毒基因组以环状游离体（Episome）形式稳定存在于宿主细胞核中，仅表达少数潜伏基因（如 EBNA1-3、LMP1/2）；在特定刺激下，病毒被激活进入裂解期（Lytic Cycle）并产生新病毒颗粒。3. 相关疾病原发感染：传染性单核细胞增多症（Infectious Mononucleosis，俗称“接吻病”）。肿瘤关联：伯基特淋巴瘤（Burkitt Lymphoma）、鼻咽癌（NPC）、霍奇金淋巴瘤、EBV 阳性胃癌及 NK/T 细胞淋巴瘤等。自身免疫病：近期多项研究证实 EBV 感染是多发性硬化症（Multiple Sclerosis, MS）发病的关键危险因素。

二、 为什么在分子生物学与基因工程中使用 EBV？EBV 独特的生物学特性（尤其是其在宿主核内保持环状游离体复制以及转化 B 淋巴细胞的能力），使其成为分子生物学领域极其关键的工具载体与实验室模型：

1. 建立人类永生化 B 细胞系（LCLs）原理：EBV 的潜伏期基因组合（如 EBNA2、LMP1 等）能持续激活宿主细胞的 NF-κB、Notch 等信号通路，驱使静息的原代 B 细胞无限增殖，形成淋巴母细胞样细胞系（Lymphoblastoid Cell Lines, LCLs）。应用价值：基因组学标准样本：人类基因组计划、HapMap 计划、千人基因组计划（1000 Genomes）中的绝大多数人类 DNA 样本（如著名的基因组金标准细胞系 GM12878）均来自于 EBV 转化的 LCL。细胞库与疾病资源保存：可从患者微量血液中提离 B 细胞并永生化，建立永久保存的基因组资源库。

2. 构建非整合型哺乳动物表达载体（Episomal Vectors）原理：EBV 拥有独特的复制起点 oriP 和反式作用蛋白 EBNA1。在包含 oriP 和 EBNA1 表达框的重组质粒（如 pCEP4、pREP4）转染宿主细胞后：EBNA1 结合 oriP，借助宿主 DNA 聚合酶在 S 期完成质粒的精准同步复制（每个细胞维持约 5–100 个拷贝）。EBNA1 借助其染色质结合域将质粒“锚定”在宿主有丝分裂期染色体上，使其随细胞分裂均匀分配到子代细胞中。主要优势：非整合性：无需插入宿主染色体，避免了随机整合导致的基因打靶破坏及位置效应（Position Effect）引发的基因沉默。高表达与可逆回收：可长期稳定表达外源基因，且能通过 Hirt 抽提等方法直接从细胞中回收完整质粒。

3. 表观遗传与三维基因组学（3D Genomics）研究模型原理：EBV 在不同的潜伏阶段（Type I、II、III 潜伏期）展现出极度严密的 DNA 甲基化、组蛋白修饰和长程染色质环化（Chromatin Looping）重构。应用价值：它是研究宿主-病毒互作、增强子-启动子高阶互作以及 3D 染色质结构演化的天然模型。

4. 基因组不稳定性与染色体断裂机制研究原理：EBNA1 蛋白不仅结合病毒自身的 oriP，还会特异性识别并结合人类基因组（如 11q23.3 区域）上的密集重复序列。在细胞 S 期，这会引发复制叉停滞、脆性断裂（CFS）及复杂结构变异（SVs）。应用价值：

三、 EBV 在分子生物学中的应用总结

| 应用领域 | 使用的 EBV 元素 / 系统 | 核心生物学优势 |
| :--- | :--- | :--- |
